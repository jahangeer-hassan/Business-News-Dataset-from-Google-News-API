## Publisher Analysis

##  Which publishers published the highest number of news articles?

```
df["publisher"].value_counts()
```

## Top Publishers Bar Chart

```
df["publisher"].value_counts().plot(kind="bar")

plt.xlabel("Publisher")
plt.ylabel("Number of News")
plt.title("Top News Publishers")

plt.show()
```

## Title Analysis

##  How many unique news titles are in the dataset?

```
df["title"].nunique()
```

## What is the distribution of news title lengths?

```
df["title_length"] = df["title"].str.len()

df["title_length"].describe()
```

## Title Length Distribution Histogram

```
df["title_length"].hist(bins=15)

plt.xlabel("Title Length")
plt.ylabel("Frequency")
plt.title("Title Length Distribution")

plt.show()
```

## Snippet Analysis

##  What is the distribution of news snippet lengths?

```
df["snippet_length"] = df["snippet"].str.len()

df["snippet_length"].describe()
```

## Snippet Length Distribution Histogram

```
df["snippet_length"].hist(bins=15)

plt.xlabel("Snippet Length")
plt.ylabel("Frequency")
plt.title("Snippet Length Distribution")

plt.show()
```

## Timestamp Analysis

##  How many news articles were published each day?

```
df["date"] = pd.to_datetime(df["timestamp"]).dt.date

df["date"].value_counts()
```

## News Publishing Trend Line Chart

```
df["date"].value_counts().sort_index().plot(kind="line")

plt.xlabel("Date")
plt.ylabel("Number of News")
plt.title("News Publishing Trend")

plt.show()
```

## Relationship Analysis

##  Is there any relationship between title length and snippet length?

## Title Length vs Snippet Length Scatter Plot

```
plt.scatter(df["title_length"], df["snippet_length"])

plt.xlabel("Title Length")
plt.ylabel("Snippet Length")
plt.title("Title Length vs Snippet Length")

plt.show()
```

## What patterns can be observed from the correlation heatmap?

## Correlation Heatmap

```
import seaborn as sns

sns.heatmap(df.corr(numeric_only=True), annot=True)

plt.title("Correlation Heatmap")

plt.show()
```
